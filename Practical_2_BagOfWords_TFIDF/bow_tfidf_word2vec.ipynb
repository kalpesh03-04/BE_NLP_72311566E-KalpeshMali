```python
import pandas as pd

df = pd.read_csv("Tweets.csv")

df.head()
```

```python
text_data = df["text"]
text_data.head()
```

```python
from sklearn.feature_extraction.text import CountVectorizer

vectorizer = CountVectorizer()

X = vectorizer.fit_transform(text_data)

print("Vocabulary:")
print(vectorizer.get_feature_names_out())

print("Bag of Words Matrix:")
print(X.toarray())
```

```python
import numpy as np

bow_matrix = X.toarray()

normalized_bow = bow_matrix / bow_matrix.sum(axis=1, keepdims=True)

print("Normalized Bag of Words:")
print(normalized_bow)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer()

X_tfidf = tfidf.fit_transform(text_data)

print("TF-IDF Features:")
print(tfidf.get_feature_names_out())

print("TF-IDF Matrix:")
print(X_tfidf.toarray())
```

```python
from gensim.models import Word2Vec

sentences = [text.split() for text in text_data]

model = Word2Vec(sentences, vector_size=50, window=3, min_count=1)

print("Word vector example:")
print(model.wv[model.wv.index_to_key[0]])
```
